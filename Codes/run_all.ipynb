```python
EPOCHS = 30
# seeds = [42, 33, 21]
seeds = [42]
```

```python
# datasets = ["CIFAR10", "CIFAR100", "CINIC10", "SVHN", "TinyImageNet", "PathMNIST", "TissueMNIST", "BloodMNIST", "OrganAMNIST"]
datasets = ["CIFAR10"]
# models = ["PreActResNet18", "ResNet18", "WideResNet28", "SENet18", "VitSmall", "VitBase", "ViT"]
models = ["ViTCIFAR10"]
attacks = ["SORA", "FGSM", "FGSM-RS", "GradAlign", "NFGSM", "AAER", "ZeroGrad", "MultiGrad", "ATAS", "PGD", "ELLE", "TRADES"]
# attacks = ["FGSM-RS", "MultiGrad", "GradAlign", "ELLE"]
# root_path = "results"
root_path = "results2"
```

```python
for seed in seeds:
    for dataset in datasets:
        for model in models:
            for attack in attacks:
                print(f"Evaluating '{dataset}' dataset, '{model}' model, '{attack}' attack.")
                if dataset in ["PathMNIST", "TissueMNIST", "BloodMNIST", "OrganAMNIST"]:
                    if dataset in ["TissueMNIST"]:
                        EXTRA_CONFIG = " --scheduler CosineAnnealing --initial_lr 0.05 --epsilon 3"
                    else:
                        EXTRA_CONFIG = " --scheduler CosineAnnealing --initial_lr 0.05"
                elif dataset in ["CIFAR10", "CIFAR100", "TinyImageNet"]:
                    if model in ["ViT", "ViTCIFAR10"]:
                        EXTRA_CONFIG = "--batch_size 128 --initial_lr 5e-5 --optimizer AdamW --scheduler WarmupLambda" 
                    else:
                        EXTRA_CONFIG = ""
                # output = os.popen(f"python main.py --root_path {root_path} --dataset {dataset} --model {model} --attack {attack} --epochs {EPOCHS}").read()
                output = subprocess.check_output(f"python main.py --root_path {root_path} --dataset {dataset} --model {model} --attack {attack} --epochs {EPOCHS} --seed {seed} --track_alignment --normalize_dataset {EXTRA_CONFIG}", shell=True, text=True)
                print(output)
                
                print("=" * 50)
```
